```python
import pandas as pd
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split

from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
```

```python
df=pd.read_csv(r"C:\Users\khanr\OneDrive\Desktop\New folder\application_train.csv")
print(df.head())
```

```python
df.isna().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
categorical_cols = df.select_dtypes(include=['object', 'category']).columns

for col in categorical_cols:
    df[col]=df[col].astype(str)
    df[col] = le.fit_transform(df[col])
print(df.head())
```

```python
X=df.drop(columns=['TARGET'])
y=df['TARGET']


```

```python
print("Dataset Shape;", X.shape)
print("Target Shape:", y.shape)
print("\nFirts 5Rows:")
print(X.head())
print("\nClass Distribution:")
print(y.value_counts())
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("\nTraining Data:", X_train.shape)
print("Testing Data :", X_test.shape)
```

```python
from sklearn.impute import SimpleImputer

imputer = SimpleImputer(strategy='median')

X_train = imputer.fit_transform(X_train)
X_test = imputer.transform(X_test)
```

```python
from imblearn.over_sampling import SMOTE

smote = SMOTE(random_state=42)

X_resampled, y_resampled = smote.fit_resample(X_train, y_train)
```

```python
decision_tree = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

decision_tree.fit(X_train, y_train)

dt_pred = decision_tree.predict(X_test)

```

```python
random_forest = RandomForestClassifier(
    n_estimators=100,
    max_depth=5,
    random_state=42
)

random_forest.fit(X_train, y_train)

rf_pred = random_forest.predict(X_test)
```

```python
# Base model used inside Bagging
base_tree = DecisionTreeClassifier(
    random_state=42
)

bagging = BaggingClassifier(
    estimator=base_tree,
    n_estimators=100,
    max_samples=3000,
    bootstrap=True,
    random_state=42
)

bagging.fit(X_train, y_train)

bagging_pred = bagging.predict(X_test)
```

```python
# ============================================================
# 7. Evaluation Function
# ============================================================

def evaluate_model(DecisionTreeClassifier, y_true, y_pred):

    accuracy = accuracy_score(y_true, y_pred)

    precision = precision_score(
        y_true,
        y_pred,
        average="weighted"
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="weighted"
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="weighted"
    )

    print("\n" + "=" * 60)
    print(DecisionTreeClassifier)
    print("=" * 60)

    print("Accuracy :", round(accuracy, 4))
    print("Precision:", round(precision, 4))
    print("Recall   :", round(recall, 4))
    print("F1 Score :", round(f1, 4))

    print("\nConfusion Matrix:")
    print(confusion_matrix(y_true, y_pred))

    print("\nClassification Report:")
    print(classification_report(y_true, y_pred))

    return [
        DecisionTreeClassifier,
        accuracy,
        precision,
        recall,
        f1
    ]
```

```python
 #8. Evaluate Decision Tree
# ============================================================

dt_result = evaluate_model(
    "Decision Tree",
    y_test,
    dt_pred
)
```

```python
# ============================================================
# 9. Evaluate Random Forest
# ============================================================

rf_result = evaluate_model(
    "Random Forest",
    y_test,
    rf_pred
)
```

```python
# ============================================================
# 10. Evaluate Bagging
# ============================================================

bagging_result = evaluate_model(
    "Bagging Classifier",
    y_test,
    bagging_pred
)
```

```python
# ============================================================
# 11. Compare All Models
# ============================================================

results = pd.DataFrame(
    [
        dt_result,
        rf_result,
        bagging_result
    ],
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]
)

print("\n" + "=" * 60)
print("FINAL MODEL COMPARISON")
print("=" * 60)

print(results.round(4))
```

```python
best_model = results.loc[
    results["Accuracy"].idxmax()
]

print("\nBest Model:", best_model["Model"])
print("Best Accuracy:", round(best_model["Accuracy"], 4))
```
