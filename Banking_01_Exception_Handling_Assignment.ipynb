### Q33. Process transfers while isolating bad rows for banking case 1; use sample reference `ACC-1033` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
def process_transfers(account, rows):
    balance = 8000

    for row in rows:
        try:
            amount = float(row["amount"])
            if amount <= 0:
                raise ValueError("Amount must be positive")
            if amount > balance:
                raise ValueError("Insufficient balance")

            balance -= amount
            print(f"{account}: Transfer successful - {amount}")

        except (KeyError,ValueError) as e:
            print(f"{account}: Bad row skipped - {e}")

    print(f"{account}: Final balance = {balance}")


rows = [
    {"amount": "1000"},
    {"amount": "abc"},     # bad row
    {"amount": "2000"},
    {"amount": "-500"}     # bad row
]

process_transfers("ACC-1033", rows)
```

### Q34. Retry a transient payment gateway for banking case 1; use sample reference `ACC-1034` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
class GatewayError(Exception):
    pass


def process_payment(account, amount):
    for attempt in range(3):
        try:
            if attempt < 2:
                raise GatewayError("Temporary gateway failure")
            print(f"{account}: Payment successful - {amount}")
            return True
        except GatewayError as e:
            print(f"{account}: Retry {attempt + 1} - {e}")

    print(f"{account}: Payment failed after 3 attempts")
    return False


process_payment("ACC-1034", 1000)
```

### Q87. Make batch settlement failure-isolated for banking case 2; use sample reference `ACC-1087` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write# Q87 - Failure-isolated batch settlement
# Banking Case 2


# -----------------------------
# Custom business exceptions
# -----------------------------
class BankingError(Exception):
    """Base exception for expected banking/business-rule failures."""
    pass


class InvalidAccountError(BankingError):
    pass


class InvalidAmountError(BankingError):
    pass


class InsufficientFundsError(BankingError):
    pass


class SettlementError(Exception):
    """Raised when an unexpected settlement operation fails."""
    pass


# -----------------------------
# Sample bank balances
# -----------------------------
accounts = {
    "ACC-1001": 5000.00,
    "ACC-1087": 3000.00,
    "ACC-2002": 1200.00,
    "ACC-3003": 7000.00
}


def settle_transfer(from_account, to_account, amount):
    """
    Settle one transfer safely.

    Expected business-rule failures raise BankingError subclasses.
    Unexpected failures are wrapped in SettlementError while
    preserving the original cause.
    """

    # Validate account references
    if not isinstance(from_account, str) or not isinstance(to_account, str):
        raise InvalidAccountError("Account references must be strings.")

    if not from_account.strip() or not to_account.strip():
        raise InvalidAccountError("Account reference cannot be empty.")

    if from_account not in accounts:
        raise InvalidAccountError(
            f"Source account {from_account} does not exist."
        )

    if to_account not in accounts:
        raise InvalidAccountError(
            f"Destination account {to_account} does not exist."
        )

    if from_account == to_account:
        raise InvalidAccountError(
            "Source and destination accounts cannot be the same."
        )

    # Validate amount
    if isinstance(amount, bool) or not isinstance(amount, (int, float)):
        raise InvalidAmountError("Settlement amount must be numeric.")

    if amount <= 0:
        raise InvalidAmountError(
            "Settlement amount must be greater than zero."
        )

    if accounts[from_account] < amount:
        raise InsufficientFundsError(
            f"{from_account} has insufficient funds."
        )

    # Save original balances so that partial updates can be reversed
    old_source_balance = accounts[from_account]
    old_destination_balance = accounts[to_account]

    try:
        # Smallest risky settlement section
        accounts[from_account] -= amount

        # Realistic infrastructure/programming failure simulation
        if from_account == "ACC-1087":
            raise ConnectionError(
                "Core banking settlement service unavailable."
            )

        accounts[to_account] += amount

    except (ConnectionError, OSError) as exc:
        # Roll back partially changed balances
        accounts[from_account] = old_source_balance
        accounts[to_account] = old_destination_balance

        # Preserve original cause using "raise ... from ..."
        raise SettlementError(
            f"Settlement failed for {from_account}."
        ) from exc

    return {
        "from": from_account,
        "to": to_account,
        "amount": amount,
        "status": "SUCCESS"
    }


def batch_settlement(batch):
    """
    Process every transaction independently.

    One failure is recorded and the next transaction continues.
    """

    if batch is None:
        return []

    if not isinstance(batch, list):
        raise TypeError("Batch must be a list.")

    results = []

    for index, transaction in enumerate(batch, start=1):

        try:
            # Empty / invalid transaction input
            if not isinstance(transaction, dict):
                raise TypeError(
                    "Each batch item must be a dictionary."
                )

            if not transaction:
                raise ValueError(
                    "Transaction cannot be empty."
                )

            result = settle_transfer(
                transaction.get("from"),
                transaction.get("to"),
                transaction.get("amount")
            )

        # Expected banking/business-rule failures
        except BankingError as exc:
            result = {
                "transaction": index,
                "status": "REJECTED",
                "error_type": type(exc).__name__,
                "reason": str(exc)
            }

        # Invalid input/programming problem
        except (TypeError, ValueError) as exc:
            result = {
                "transaction": index,
                "status": "INVALID",
                "error_type": type(exc).__name__,
                "reason": str(exc)
            }

        # Unexpected infrastructure settlement failure
        except SettlementError as exc:
            result = {
                "transaction": index,
                "status": "FAILED",
                "error_type": type(exc).__name__,
                "reason": str(exc),
                "original_cause": str(exc.__cause__)
            }

        # Last safety boundary for this single batch item
        except Exception as exc:
            result = {
                "transaction": index,
                "status": "SYSTEM_ERROR",
                "error_type": type(exc).__name__,
                "reason": str(exc)
            }

        results.append(result)

    return results


# ------------------------------------
# Test batch
# ------------------------------------
batch = [

    # Normal settlement
    {
        "from": "ACC-1001",
        "to": "ACC-2002",
        "amount": 500
    },

    # Realistic failure path
    {
        "from": "ACC-1087",
        "to": "ACC-3003",
        "amount": 600
    },

    # Boundary value
    {
        "from": "ACC-2002",
        "to": "ACC-3003",
        "amount": 0
    },

    # Invalid type
    {
        "from": "ACC-1001",
        "to": "ACC-3003",
        "amount": "1000"
    },

    # Empty input
    {}
]


print("Balances before settlement:")
print(accounts)

results = batch_settlement(batch)

print("\nSettlement Results:")
for item in results:
    print(item)

print("\nBalances after settlement:")
print(accounts) 
```

```python
# Write # Q88 - Report exact paths in nested statements
# Banking Case 2


# ---------------------------------
# Custom banking exceptions
# ---------------------------------
class BankingError(Exception):
    """Base class for expected business-rule failures."""
    pass


class InvalidAccountError(BankingError):
    pass


class InvalidAmountError(BankingError):
    pass


class InsufficientFundsError(BankingError):
    pass


class SettlementError(Exception):
    """Unexpected banking/infrastructure failure."""
    pass


# ---------------------------------
# Sample account balances
# ---------------------------------
accounts = {
    "ACC-1001": 5000.00,
    "ACC-1088": 3000.00,
    "ACC-2002": 1500.00,
    "ACC-3003": 7000.00
}


# ---------------------------------
# Process one transfer
# ---------------------------------
def settle_transfer(from_account, to_account, amount):

    # Validate account types
    if not isinstance(from_account, str):
        raise InvalidAccountError(
            "Source account must be a string."
        )

    if not isinstance(to_account, str):
        raise InvalidAccountError(
            "Destination account must be a string."
        )

    # Empty input
    if not from_account.strip():
        raise InvalidAccountError(
            "Source account cannot be empty."
        )

    if not to_account.strip():
        raise InvalidAccountError(
            "Destination account cannot be empty."
        )

    # Account existence
    if from_account not in accounts:
        raise InvalidAccountError(
            f"Account {from_account} does not exist."
        )

    if to_account not in accounts:
        raise InvalidAccountError(
            f"Account {to_account} does not exist."
        )

    # Invalid same-account transfer
    if from_account == to_account:
        raise InvalidAccountError(
            "Source and destination accounts must be different."
        )

    # Amount validation
    if isinstance(amount, bool) or not isinstance(amount, (int, float)):
        raise InvalidAmountError(
            "Amount must be numeric."
        )

    # Boundary value
    if amount <= 0:
        raise InvalidAmountError(
            "Amount must be greater than zero."
        )

    if accounts[from_account] < amount:
        raise InsufficientFundsError(
            f"Insufficient funds in {from_account}."
        )

    # Save original balances
    old_source_balance = accounts[from_account]
    old_destination_balance = accounts[to_account]

    try:
        # First update
        accounts[from_account] -= amount

        # Realistic infrastructure failure
        if from_account == "ACC-1088":
            raise ConnectionError(
                "Core banking service unavailable."
            )

        # Second update
        accounts[to_account] += amount

    except (ConnectionError, OSError) as exc:

        # Roll back partial balance changes
        accounts[from_account] = old_source_balance
        accounts[to_account] = old_destination_balance

        # Preserve original exception
        raise SettlementError(
            f"Settlement failed for {from_account}."
        ) from exc

    return {
        "from": from_account,
        "to": to_account,
        "amount": amount,
        "status": "SUCCESS"
    }


# ---------------------------------
# Process nested statements
# ---------------------------------
def process_statements(statements):

    if statements is None:
        return []

    if not isinstance(statements, list):
        raise TypeError(
            "statements must be a list."
        )

    results = []

    for statement_index, statement in enumerate(statements):

        statement_path = f"statements[{statement_index}]"

        # Validate statement
        if not isinstance(statement, dict):
            results.append({
                "path": statement_path,
                "status": "INVALID",
                "reason": "Statement must be a dictionary."
            })
            continue

        transactions = statement.get("transactions")

        if not isinstance(transactions, list):
            results.append({
                "path": f"{statement_path}.transactions",
                "status": "INVALID",
                "reason": "transactions must be a list."
            })
            continue

        # Process every nested transaction independently
        for transaction_index, transaction in enumerate(transactions):

            transaction_path = (
                f"{statement_path}"
                f".transactions[{transaction_index}]"
            )

            try:
                if not isinstance(transaction, dict):
                    raise TypeError(
                        "Transaction must be a dictionary."
                    )

                if not transaction:
                    raise ValueError(
                        "Transaction cannot be empty."
                    )

                # Smallest risky field access
                try:
                    from_account = transaction["from"]
                except KeyError as exc:
                    raise ValueError(
                        f"Missing field at "
                        f"{transaction_path}.from"
                    ) from exc

                try:
                    to_account = transaction["to"]
                except KeyError as exc:
                    raise ValueError(
                        f"Missing field at "
                        f"{transaction_path}.to"
                    ) from exc

                try:
                    amount = transaction["amount"]
                except KeyError as exc:
                    raise ValueError(
                        f"Missing field at "
                        f"{transaction_path}.amount"
                    ) from exc

                result = settle_transfer(
                    from_account,
                    to_account,
                    amount
                )

                result["path"] = transaction_path
                results.append(result)

            # Expected banking/business-rule errors
            except BankingError as exc:

                # Identify more exact field
                if isinstance(exc, InvalidAmountError):
                    error_path = f"{transaction_path}.amount"

                elif isinstance(exc, InvalidAccountError):
                    error_path = transaction_path

                else:
                    error_path = transaction_path

                results.append({
                    "path": error_path,
                    "status": "REJECTED",
                    "error_type": type(exc).__name__,
                    "reason": str(exc)
                })

            # Invalid structure/programming input
            except (TypeError, ValueError) as exc:
                results.append({
                    "path": transaction_path,
                    "status": "INVALID",
                    "error_type": type(exc).__name__,
                    "reason": str(exc),
                    "original_cause":
                        str(exc.__cause__)
                        if exc.__cause__
                        else None
                })

            # Infrastructure failure
            except SettlementError as exc:
                results.append({
                    "path": transaction_path,
                    "status": "FAILED",
                    "error_type": type(exc).__name__,
                    "reason": str(exc),
                    "original_cause": str(exc.__cause__)
                })

    return results


# ---------------------------------
# Test nested banking statements
# ---------------------------------
statements = [

    {
        "statement_id": "STMT-001",
        "transactions": [

            # Valid transaction
            {
                "from": "ACC-1001",
                "to": "ACC-2002",
                "amount": 500
            },

            # Boundary value
            {
                "from": "ACC-1001",
                "to": "ACC-3003",
                "amount": 0
            }
        ]
    },

    {
        "statement_id": "STMT-002",
        "transactions": [

            # Realistic infrastructure failure
            {
                "from": "ACC-1088",
                "to": "ACC-3003",
                "amount": 600
            },

            # Invalid type
            {
                "from": "ACC-2002",
                "to": "ACC-3003",
                "amount": "700"
            },

            # Empty transaction
            {}
        ]
    }
]


print("Balances before:")
print(accounts)

results = process_statements(statements)

print("\nResults:")

for result in results:
    print(result)

print("\nBalances after:")
print(accounts)
```

### Q89. Design idempotent recovery after timeout for banking case 2; use sample reference `ACC-1089` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
# Q89 - Idempotent recovery after timeout
# Banking Case 2


# -----------------------------------
# Custom banking exceptions
# -----------------------------------
class BankingError(Exception):
    """Base class for expected business-rule failures."""
    pass


class InvalidAccountError(BankingError):
    pass


class InvalidAmountError(BankingError):
    pass


class InsufficientFundsError(BankingError):
    pass


class SettlementError(Exception):
    """Unexpected infrastructure/settlement failure."""
    pass


# -----------------------------------
# Sample account balances
# -----------------------------------
accounts = {
    "ACC-1089": 5000.00,
    "ACC-2001": 2000.00,
    "ACC-3001": 4000.00
}


# Stores completed settlements using
# transaction/reference ID as idempotency key.
settlement_store = {}


# -----------------------------------
# Validation
# -----------------------------------
def validate_transfer(reference, from_account, to_account, amount):

    # Empty reference
    if not isinstance(reference, str):
        raise TypeError("Reference must be a string.")

    if not reference.strip():
        raise ValueError("Reference cannot be empty.")

    # Account validation
    if not isinstance(from_account, str):
        raise InvalidAccountError(
            "Source account must be a string."
        )

    if not isinstance(to_account, str):
        raise InvalidAccountError(
            "Destination account must be a string."
        )

    if not from_account.strip() or not to_account.strip():
        raise InvalidAccountError(
            "Account number cannot be empty."
        )

    if from_account not in accounts:
        raise InvalidAccountError(
            f"Source account {from_account} does not exist."
        )

    if to_account not in accounts:
        raise InvalidAccountError(
            f"Destination account {to_account} does not exist."
        )

    if from_account == to_account:
        raise InvalidAccountError(
            "Source and destination accounts must be different."
        )

    # Invalid amount type
    if isinstance(amount, bool) or not isinstance(amount, (int, float)):
        raise InvalidAmountError(
            "Amount must be numeric."
        )

    # Boundary value
    if amount <= 0:
        raise InvalidAmountError(
            "Amount must be greater than zero."
        )

    if accounts[from_account] < amount:
        raise InsufficientFundsError(
            f"Insufficient funds in {from_account}."
        )


# -----------------------------------
# Core settlement
# -----------------------------------
def perform_settlement(
    reference,
    from_account,
    to_account,
    amount,
    simulate_timeout=False
):
    """
    Process one settlement idempotently.

    If the same reference has already succeeded,
    return the previous result instead of applying
    the transfer again.
    """

    validate_transfer(
        reference,
        from_account,
        to_account,
        amount
    )

    # -----------------------------------
    # Idempotency check
    # -----------------------------------
    if reference in settlement_store:

        old = settlement_store[reference]

        # Same reference must describe
        # exactly the same transaction.
        if (
            old["from"] != from_account
            or old["to"] != to_account
            or old["amount"] != amount
        ):
            raise BankingError(
                f"Reference {reference} was already used "
                "for a different transaction."
            )

        return {
            **old,
            "recovered": True,
            "message": "Previously completed settlement returned."
        }

    # Save balances in case an update fails
    old_source_balance = accounts[from_account]
    old_destination_balance = accounts[to_account]

    try:
        # -----------------------------------
        # Apply transfer
        # -----------------------------------
        accounts[from_account] -= amount

        accounts[to_account] += amount

        # Store durable successful result BEFORE
        # returning the response.
        result = {
            "reference": reference,
            "from": from_account,
            "to": to_account,
            "amount": amount,
            "status": "COMPLETED"
        }

        settlement_store[reference] = result.copy()

        # -----------------------------------
        # Realistic failure:
        # transfer completed, but response lost
        # -----------------------------------
        if simulate_timeout:
            raise TimeoutError(
                "Settlement completed but response timed out."
            )

        return {
            **result,
            "recovered": False
        }

    except TimeoutError:
        # IMPORTANT:
        # Do NOT roll back here.
        #
        # The transfer may already be committed.
        # Recovery will check the idempotency record.
        raise

    except (OSError, ConnectionError) as exc:

        # If failure occurred before successful durable
        # completion, protect partially updated balances.
        accounts[from_account] = old_source_balance
        accounts[to_account] = old_destination_balance

        settlement_store.pop(reference, None)

        raise SettlementError(
            f"Settlement failed for {reference}."
        ) from exc

    except Exception:
        # Protect partially updated balances for
        # unexpected local failures.
        if reference not in settlement_store:
            accounts[from_account] = old_source_balance
            accounts[to_account] = old_destination_balance

        raise


# -----------------------------------
# Timeout-safe recovery
# -----------------------------------
def settle_with_recovery(
    reference,
    from_account,
    to_account,
    amount,
    simulate_timeout=False
):

    try:
        return perform_settlement(
            reference,
            from_account,
            to_account,
            amount,
            simulate_timeout
        )

    except TimeoutError as exc:

        print(
            f"Timeout received for {reference}. "
            "Checking settlement status..."
        )

        try:
            # Retry using EXACTLY the same
            # idempotency/reference key.
            recovered = perform_settlement(
                reference,
                from_account,
                to_account,
                amount,
                simulate_timeout=False
            )

            recovered["original_timeout"] = str(exc)

            return recovered

        except BankingError:
            raise

        except Exception as recovery_exc:
            raise SettlementError(
                f"Could not safely recover settlement {reference}."
            ) from recovery_exc


# -----------------------------------
# Edge-case tests
# -----------------------------------

def run_test(name, func):
    print(f"\n--- {name} ---")

    try:
        print(func())

    except BankingError as exc:
        print(
            "BUSINESS ERROR:",
            type(exc).__name__,
            "-",
            exc
        )

    except (TypeError, ValueError) as exc:
        print(
            "INPUT ERROR:",
            type(exc).__name__,
            "-",
            exc
        )

    except SettlementError as exc:
        print(
            "SYSTEM ERROR:",
            exc
        )

        if exc.__cause__:
            print(
                "Original cause:",
                type(exc.__cause__).__name__,
                "-",
                exc.__cause__
            )


print("Balances before:")
print(accounts)


# 1. Empty input
run_test(
    "Empty reference",
    lambda: settle_with_recovery(
        "",
        "ACC-1089",
        "ACC-2001",
        100
    )
)


# 2. Invalid type
run_test(
    "Invalid amount type",
    lambda: settle_with_recovery(
        "TXN-TYPE-001",
        "ACC-1089",
        "ACC-2001",
        "500"
    )
)


# 3. Boundary value
run_test(
    "Zero amount",
    lambda: settle_with_recovery(
        "TXN-ZERO-001",
        "ACC-1089",
        "ACC-2001",
        0
    )
)


# 4. Realistic timeout path
run_test(
    "Timeout recovery",
    lambda: settle_with_recovery(
        "TXN-1089",
        "ACC-1089",
        "ACC-2001",
        1000,
        simulate_timeout=True
    )
)


print("\nBalances after timeout recovery:")
print(accounts)


# 5. Retry SAME settlement again
run_test(
    "Retry same reference",
    lambda: settle_with_recovery(
        "TXN-1089",
        "ACC-1089",
        "ACC-2001",
        1000
    )
)


print("\nBalances after retry:")
print(accounts)
```

### Q95. Create a transaction context manager for banking case 3; use sample reference `ACC-1095` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
from contextlib import contextmanager


class BankingError(Exception):
    pass


class TransactionError(Exception):
    pass


@contextmanager
def transaction(accounts, src, dest):
    old_src = accounts[src]
    old_dest = accounts[dest]

    try:
        yield
    except Exception as e:
        accounts[src] = old_src
        accounts[dest] = old_dest
        raise TransactionError(
            "ACC-1095: Transaction rolled back"
        ) from e


def transfer(accounts, src, dest, amount):
    ref = "ACC-1095"

    if not src or not dest:
        raise BankingError(f"{ref}: Empty account")

    try:
        amount = float(amount)
    except (TypeError, ValueError) as e:
        raise BankingError(f"{ref}: Invalid amount") from e

    if amount <= 0:
        raise BankingError(f"{ref}: Amount must be > 0")

    if src not in accounts or dest not in accounts:
        raise BankingError(f"{ref}: Account not found")

    if accounts[src] < amount:
        raise BankingError(f"{ref}: Insufficient balance")

    with transaction(accounts, src, dest):
        accounts[src] -= amount
        accounts[dest] += amount

    return f"{ref}: Transfer successful"


accounts = {"A1": 5000, "A2": 2000}

try:
    print(transfer(accounts, "A1", "A2", 1000))
    print(accounts)
except Exception as e:
    print(e)
```

### Q97. Make batch settlement failure-isolated for banking case 3; use sample reference `ACC-1097` and explain the result.

2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
class BankingError(Exception):
    pass

class SettlementError(Exception):
    pass


def settle_one(accounts, tx):
    ref = "ACC-1097"

    if not tx:
        raise BankingError(f"{ref}: Empty transaction")

    try:
        src = tx["from"]
        dest = tx["to"]
        amount = float(tx["amount"])
    except (KeyError, TypeError, ValueError) as e:
        raise BankingError(f"{ref}: Invalid transaction") from e

    if amount <= 0:
        raise BankingError(f"{ref}: Amount must be > 0")

    if src not in accounts or dest not in accounts:
        raise BankingError(f"{ref}: Account not found")

    if accounts[src] < amount:
        raise BankingError(f"{ref}: Insufficient balance")

    old_src, old_dest = accounts[src], accounts[dest]

    try:
        accounts[src] -= amount

        if tx.get("fail"):
            raise ConnectionError("Settlement service failed")

        accounts[dest] += amount

    except ConnectionError as e:
        accounts[src] = old_src
        accounts[dest] = old_dest
        raise SettlementError(
            f"{ref}: Settlement rolled back"
        ) from e


def batch_settle(accounts, transactions):
    results = []

    for tx in transactions:
        try:
            settle_one(accounts, tx)
            results.append("SUCCESS")
        except Exception as e:
            results.append(f"FAILED: {e}")

    return results


accounts = {"A1": 5000, "A2": 2000, "A3": 1000}

transactions = [
    {"from": "A1", "to": "A2", "amount": 1000},
    {"from": "A2", "to": "A3", "amount": 500, "fail": True},
    {"from": "A1", "to": "A3", "amount": 500}
]

print(batch_settle(accounts, transactions))
print(accounts)
```

### Q99. Design idempotent recovery after timeout for banking case 3; use sample reference `ACC-1099` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
class BankingError(Exception):
    pass

class RecoveryError(Exception):
    pass


processed = {}


def transfer(accounts, tx, timeout=False):
    ref = "ACC-1099"

    if not tx:
        raise BankingError(f"{ref}: Empty input")

    try:
        tx_id = tx["id"]
        src = tx["from"]
        dest = tx["to"]
        amount = float(tx["amount"])
    except (KeyError, TypeError, ValueError) as e:
        raise BankingError(f"{ref}: Invalid transaction") from e

    # Idempotent recovery
    if tx_id in processed:
        return f"{ref}: Already processed"

    if amount <= 0:
        raise BankingError(f"{ref}: Amount must be > 0")

    if src not in accounts or dest not in accounts:
        raise BankingError(f"{ref}: Account not found")

    if accounts[src] < amount:
        raise BankingError(f"{ref}: Insufficient balance")

    old_src, old_dest = accounts[src], accounts[dest]

    try:
        accounts[src] -= amount
        accounts[dest] += amount

        processed[tx_id] = True

        if timeout:
            raise TimeoutError("Response timed out")

        return f"{ref}: Transfer successful"

    except TimeoutError as e:
        # Transaction may already be completed.
        raise RecoveryError(
            f"{ref}: Timeout - retry safely with same transaction ID"
        ) from e

    except Exception as e:
        accounts[src] = old_src
        accounts[dest] = old_dest
        raise RecoveryError(f"{ref}: Transfer failed") from e


accounts = {"A1": 5000, "A2": 2000}

tx = {
    "id": "TX100",
    "from": "A1",
    "to": "A2",
    "amount": 1000
}

try:
    transfer(accounts, tx, timeout=True)
except RecoveryError as e:
    print(e)

# Safe retry
print(transfer(accounts, tx))
print(accounts)
```
