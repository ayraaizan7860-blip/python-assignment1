```python

import math

p = float(input("Enter principal amount: "))
r = float(input("Enter annual interest rate (%): "))
t = float(input("Enter time (years): "))

amount = p * math.pow((1 + r / 100), t)
compound_interest = amount - p

print("Amount =", amount)
print("Compound Interest =", compound_interest)
```

```python
import math

print("===== Scientific Calculator =====")
print("1. Square Root")
print("2. Factorial")
print("3. Logarithm (Base 10)")
print("4. Natural Log")
print("5. Power")
print("6. Sine")
print("7. Cosine")
print("8. Tangent")
print("9. GCD")
print("10. LCM")

choice = int(input("Enter your choice (1-10):"))

if choice == 1:
    n = float(input("Enter a number: "))
    print("Square Root =", math.sqrt(n))

elif choice == 2:
    n = int(input("Enter a number: "))
    print("Factorial =", math.factorial(n))

elif choice == 3:
    n = float(input("Enter a number: "))
    print("Log10 =", math.log10(n))

elif choice == 4:
    n = float(input("Enter a number: "))
    print("Natural Log =", math.log(n))

elif choice == 5:
    a = float(input("Enter base: "))
    b = float(input("Enter exponent: "))
    print("Power =", math.pow(a, b))

elif choice == 6:
    angle = float(input("Enter angle in degrees: "))
    print("Sin =", math.sin(math.radians(angle)))

elif choice == 7:
    angle = float(input("Enter angle in degrees: "))
    print("Cos =", math.cos(math.radians(angle)))

elif choice == 8:
    angle = float(input("Enter angle in degrees: "))
    print("Tan =", math.tan(math.radians(angle)))

elif choice == 9:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print("GCD =", math.gcd(a, b))

elif choice == 10:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))
    print("LCM =", math.lcm(a, b))

else:
    print("Invalid Choice")
```
