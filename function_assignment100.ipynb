Assignment 3: Calculator Using Functions
Create four separate functions:

add()
subtract()
multiply()
divide()
Requirements
Accept two numbers from the user.
Ask the user to select an operation.
Call the appropriate function.
Handle division by zero.
Print the calculated result.

```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error! Division by zero is not allowed."
    return a / b


# Accept input from the user
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

# Ask the user to choose an operation
print("\nSelect an operation:")
print("1. Add")
print("2. Subtract")
print("3. Multiply")
print("4. Divide")

choice = input("Enter your choice (1/2/3/4): ")

# Call the appropriate function
if choice == "1":
    result = add(num1, num2)
    print("Result:", result)

elif choice == "2":
    result = subtract(num1, num2)
    print("Result:", result)

elif choice == "3":
    result = multiply(num1, num2)
    print("Result:", result)

elif choice == "4":
    result = divide(num1, num2)
    print("Result:", result)

else:
    print("Invalid choice! Please select a valid operation.")
```

Write a function named calculate_factorial().

Requirements
Accept a non-negative integer.
Calculate its factorial using a loop.
Do not use math.factorial().
Return the factorial.
Handle negative input.

```python
def calculate_factorial(number):
    if number < 0:
        return "Error! Factorial is not defined for negative numbers."

    factorial = 1
    for i in range(1, number + 1):
        factorial *= i

    return factorial

num = int(input("Enter a non-negative integer: "))
result = calculate_factorial(num)
print("Factorial:", result)
```

Assignment 17: Palindrome Checker
Write a function named is_palindrome().

Requirements
Accept a string.
Ignore spaces and letter case.
Return True when the string is a palindrome.
Print a meaningful result.

```python
def is_palindrome(text):
    cleaned_text = text.replace(" ", "").lower()
    if cleaned_text == cleaned_text[::-1]:
        return True
    else:
        return False
string = input("Enter a string: ")
if is_palindrome(string):
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```

ssignment 19: Find Maximum in a List
Write a function named find_list_maximum().

Requirements
Accept a list of numbers.
Do not use max().
Return the largest list element.
Handle an empty list by returning an appropriate message.

```python
def find_list_maximum(numbers):
    if len(numbers) == 0:
        return "Error! The list is empty."

    largest = numbers[0]

    for num in numbers:
        if num > largest:
            largest = num

    return largest
numbers = list(map(int, input("Enter numbers separated by spaces: ").split()))

result = find_list_maximum(numbers)
print("Largest element:", result)
```

Assignment 20: Count Positive, Negative and Zero Values
Write a function named count_number_types().

Requirements
Accept a list of integers.
Count positive numbers, negative numbers, and zeros.
Return all three counts.
Use tuple unpacking to receive the returned values.

```python
def count_number_type(numbers):
  positive_count=0
  negative_count=0
  zero_count=0
  for num in numbers:
    if num>0:
      positive_count+=1
    elif num<0:
      negative_count+=1
    else:
        zero_count+=1
    return positive_count,negative_count,zero_count
numbers=list(map (int,(input("enter the integer separated by spaces:").split())))
positive ,negative ,zero=  count_number_type(numbers)
print("Positive numbers:", positive)
print("Negative numbers:", negative)
print("Zeros:", zero)
```
