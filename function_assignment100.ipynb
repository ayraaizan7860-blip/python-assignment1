Write a function named calculate_factorial().

Requirements
Accept a non-negative integer.
Calculate its factorial using a loop.
Do not use math.factorial().
Return the factorial.
Handle negative input.

```python
def calculate_factorial(number):
    if number < 0:
        return "Error! Factorial is not defined for negative numbers."

    factorial = 1
    for i in range(1, number + 1):
        factorial *= i

    return factorial

num = int(input("Enter a non-negative integer: "))
result = calculate_factorial(num)
print("Factorial:", result)
```

Assignment 32: Function Returning a Dictionary
Write a function named analyze_text().

Requirements
Accept a string.
Return a dictionary containing:
Total characters
Total words
Total vowels
Total digits
Total spaces
Display every result clearly.

```python
def analyze_text(text):
    vowels = "aeiouAEIOU"

    total_characters = len(text)
    total_words = len(text.split())
    total_vowels = 0
    total_digits = 0
    total_spaces = 0

    for char in text:
        if char in vowels:
            total_vowels += 1
        if char.isdigit():
            total_digits += 1
        if char == " ":
            total_spaces += 1
    return {
        "Total Characters": total_characters,
        "Total Words": total_words,
        "Total Vowels": total_vowels,
        "Total Digits": total_digits,
        "Total Spaces": total_spaces
    }


text = "Python Programming 2026"
result = analyze_text(text)

print("Text:", text)
print("\nAnalysis Result:")
for key, value in result.items():
    print(f"{key}: {value}")
```

```python
def get_even_numbers(start, end):
    if start >= end:
        print("Error: Starting number must be smaller than the ending number.")
        return []

    even_numbers = []
    for num in range(start, end + 1):
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers

result = get_even_numbers(40, 50)

print("Even Numbers:", result)
```

Accept one integer.
Return True when the number is prime.
Return False otherwise.
Use the function to print all prime numbers from 1 to 100.
Hint
A prime number has exactly two factors.



```python
def is_prime(number):

    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True

print("Prime numbers from 1 to 100:")

for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

Assignment 35: Fibonacci Function
Write a function named generate_fibonacci().

Requirements
Accept the number of terms.
Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.

```python
def generate_fibonacci(terms):
    if terms <= 0:
        return []
    if terms == 1:
        return [0]
    fibonacci = [0, 1]


    while len(fibonacci) < terms:
        fibonacci.append(fibonacci[-1] + fibonacci[-2])

    return fibonacci

result = generate_fibonacci(7)

print("Fibonacci Series:", result)

print("Zero Terms:", generate_fibonacci(0))

print("Negative Terms:", generate_fibonacci(-5))
```
