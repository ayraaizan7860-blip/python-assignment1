Do not use math.factorial().
Return the factorial.
Handle negative input.

```python
def calculate_factorial(number):
    if number < 0:
        return "Error! Factorial is not defined for negative numbers."

    factorial = 1
    for i in range(1, number + 1):
        factorial *= i

    return factorial

num = int(input("Enter a non-negative integer: "))
result = calculate_factorial(num)
print("Factorial:", result)
```

Assignment 32: Function Returning a Dictionary
Write a function named analyze_text().

Requirements
Accept a string.
Return a dictionary containing:
Total characters
Total words
Total vowels
Total digits
Total spaces
Display every result clearly.

```python
def analyze_text(text):
    vowels = "aeiouAEIOU"

    total_characters = len(text)
    total_words = len(text.split())
    total_vowels = 0
    total_digits = 0
    total_spaces = 0

    for char in text:
        if char in vowels:
            total_vowels += 1
        if char.isdigit():
            total_digits += 1
        if char == " ":
            total_spaces += 1
    return {
        "Total Characters": total_characters,
        "Total Words": total_words,
        "Total Vowels": total_vowels,
        "Total Digits": total_digits,
        "Total Spaces": total_spaces
    }


text = "Python Programming 2026"
result = analyze_text(text)

print("Text:", text)
print("\nAnalysis Result:")
for key, value in result.items():
    print(f"{key}: {value}")
```

Return False otherwise.

Hint
A prime number has exactly two factors.



```python
def is_prime(number):

    if number < 2:
        return False

    for i in range(2, number):
        if number % i == 0:
            return False

    return True

print("Prime numbers from 1 to 100:")

for num in range(1, 101):
    if is_prime(num):
        print(num, end=" ")
```

Assignment 35: Fibonacci Function
Write a function named generate_fibonacci().

Requirements
Accept the number of terms.
Return the complete Fibonacci series as a list.
Handle zero and negative term counts.
Do not use recursion in this assignment.

```python
def generate_fibonacci(terms):
    if terms <= 0:
        return []
    if terms == 1:
        return [0]
    fibonacci = [0, 1]


    while len(fibonacci) < terms:
        fibonacci.append(fibonacci[-1] + fibonacci[-2])

    return fibonacci

result = generate_fibonacci(7)

print("Fibonacci Series:", result)

print("Zero Terms:", generate_fibonacci(0))

print("Negative Terms:", generate_fibonacci(-5))
```

Assignment 92: Closure-Based Bank Balance
Write a function named create_bank_account().

Requirements

Define inner functions for:
Deposit
Withdrawal
Balance enquiry
Use nonlocal to update the balance.
Reject withdrawals greater than the available balance.
Return all three inner functions.
Hint
Return multiple functions:

return deposit, withdraw, check_balance

```python
def create_bank_account(initial_balance=0):
    balance = initial_balance

    # Deposit function
    def deposit(amount):
        nonlocal balance
        if amount > 0:
            balance += amount
            print(f"Deposited: {amount}")
        else:
            print("Invalid deposit amount.")

    # Withdrawal function
    def withdraw(amount):
        nonlocal balance
        if amount <= balance:
            balance -= amount
            print(f"Withdrawn: {amount}")
        else:
            print("Insufficient balance.")

    # Balance enquiry function
    def check_balance():
        print(f"Current Balance: {balance}")

    # Return all three functions
    return deposit, withdraw, check_balance


# Create a bank account with an initial balance
deposit, withdraw, check_balance = create_bank_account(1000)

# Perform operations
check_balance()

deposit(500)
check_balance()

withdraw(300)
check_balance()

withdraw(1500)   # Exceeds available balance
check_balance()
```

ssignment 93: Function Decorator
Create a decorator named execution_logger().

The decorator should print a message before the original function executes.
It should print another message after execution.
Apply it to an addition function.
Preserve the original function's return value.
Expected Pattern
Starting function...
Result: 30
Function completed.

```python
def execution_logger(func):

    def wrapper(a, b):
        print("Starting function...")

        # Call the original function
        result = func(a, b)

        print("Function completed.")

        # Return the original function's result
        return result

    return wrapper

# Apply the decorator
@execution_logger
def add(a, b):
    return a + b

# Accept input
num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

# Call the decorated function
result = add(num1, num2)

# Display the result
print("Result:", result)
```
